# Fongbe Challenge — Téléchargement des données audio vers Drive

Télécharge les **22 archives .tar (~33 GB)** depuis le bucket public GCS
directement vers Google Drive. À exécuter **une seule fois**.

Reprend automatiquement les téléchargements interrompus (`wget -c`).
Les fichiers déjà présents et complets sont ignorés.

```python
from pathlib import Path
from google.colab import drive
import os, subprocess

drive.mount('/content/drive')

DEST = Path('/content/drive/MyDrive/fongbe-challenge/audio')
DEST.mkdir(parents=True, exist_ok=True)
BASE_URL = 'https://storage.googleapis.com/fongbe'

FILES = [
    # test (~2.5 GB)
    'audio_test_000.tar', 'audio_test_001.tar',
    # train_reliable (~12.8 GB)
    'audio_train_reliable_000.tar', 'audio_train_reliable_001.tar',
    'audio_train_reliable_002.tar', 'audio_train_reliable_003.tar',
    'audio_train_reliable_004.tar', 'audio_train_reliable_005.tar',
    # train_religious (~4.6 GB)
    'audio_train_religious_000.tar', 'audio_train_religious_001.tar',
    'audio_train_religious_002.tar',
    # train_uncertain (~22 GB)
    'audio_train_uncertain_000.tar', 'audio_train_uncertain_001.tar',
    'audio_train_uncertain_002.tar', 'audio_train_uncertain_003.tar',
    'audio_train_uncertain_004.tar', 'audio_train_uncertain_005.tar',
    'audio_train_uncertain_006.tar', 'audio_train_uncertain_007.tar',
    'audio_train_uncertain_008.tar', 'audio_train_uncertain_009.tar',
    'audio_train_uncertain_010.tar',
]

print(f'📥 Destination : {DEST}')
print(f'{len(FILES)} archives, ~33 GB au total')
```

```python
# Téléchargement avec reprise sur interruption
# Le notebook peut être interrompu et relancé : wget -c reprend où il en était.
done = skipped = 0
for name in FILES:
    target = DEST / name
    url = f'{BASE_URL}/{name}'
    print(f'--- {name}')
    subprocess.run(['wget', '-c', '-q', '--show-progress', '-O', str(target), url], check=True)
    done += 1
    print(f'✓ {done}/{len(FILES)}')

print('✅ Tous les téléchargements sont terminés.')
```

```python
# Vérification : liste des archives et tailles
for p in sorted(DEST.glob('*.tar')):
    print(f'{p.name:35s} {p.stat().st_size/1e9:6.2f} GB')

total = sum(p.stat().st_size for p in DEST.glob('*.tar'))
print(f'\nTotal : {total/1e9:.1f} GB — attendu ~33 GB')
```
